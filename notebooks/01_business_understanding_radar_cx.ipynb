## 1. Contexto de negócio

Empresas varejistas recebem reclamações por diferentes canais, produtos e problemas. Entretanto, o volume isolado de reclamações não mostra onde a experiência do cliente está realmente se deteriorando. Para decidir onde atuar, lideranças de CX e operações precisam combinar **volume, tempo de resposta, cumprimento do SLA, resolução e satisfação**.

Sem uma visão integrada desses indicadores, a priorização tende a ser reativa: problemas recorrentes permanecem sem tratamento estrutural, equipes concentram esforços apenas nos casos mais visíveis e o impacto da demora na resposta sobre a satisfação não é mensurado.

## 2. Problema de negócio

**Como transformar reclamações públicas de consumidores em um sistema de priorização que mostre quais problemas, empresas e canais do varejo combinam alto volume, baixo índice de solução, descumprimento do SLA e baixa satisfação?**

## 3. Objetivo geral

Construir um diagnóstico orientado por dados e um dashboard executivo capaz de:

- monitorar os principais KPIs de atendimento ao consumidor;
- identificar os maiores motivos de reclamação no varejo;
- medir o cumprimento do prazo máximo de resposta;
- analisar a relação entre agilidade, resolução e satisfação;
- comparar o desempenho entre empresas, segmentos e canais;
- priorizar problemas por impacto e urgência;
- recomendar ações para CX e operações.

## 7. KPIs e regras de cálculo

| KPI | Definição inicial | Base de cálculo |
|---|---|---|
| Total de reclamações | Contagem de registros válidos | Todas as reclamações do recorte |
| Taxa de resposta | Reclamações respondidas ÷ total de reclamações | Campo `Respondida` |
| Tempo médio de resposta | Média de `Tempo Resposta` | Apenas respostas com tempo válido |
| Tempo mediano de resposta | Mediana de `Tempo Resposta` | Apenas respostas com tempo válido |
| Cumprimento do SLA | Respostas em até 10 dias ÷ respostas com tempo válido | `Tempo Resposta <= 10` |
| Taxa de resolução observada | Resolvidas ÷ reclamações avaliadas como resolvidas ou não resolvidas | `Avaliação Reclamação` |
| Índice de solução oficial | Regra metodológica oficial, que considera reclamações não avaliadas como resolvidas | Situação e avaliação |
| Satisfação média | Média das notas de 1 a 5 | Apenas `Nota do Consumidor` válida |
| Taxa de baixa satisfação | Notas 1 e 2 ÷ total de notas válidas | Reclamações avaliadas |
| Participação no volume | Reclamações da categoria ÷ total do recorte | Segmento, empresa ou problema |

> As duas métricas de solução serão mantidas separadas para evitar que reclamações não avaliadas pelo consumidor sejam interpretadas como resolução explicitamente confirmada.

## 8. Hipóteses analíticas

- **H1:** reclamações respondidas dentro do SLA apresentam maior satisfação média.
- **H2:** o tempo de resposta é maior entre reclamações não resolvidas.
- **H3:** poucos grupos de problema concentram a maior parte das reclamações do varejo (efeito Pareto).
- **H4:** problemas relacionados a cobrança, entrega e reembolso apresentam maior taxa de baixa satisfação.
- **H5:** empresas com menor cumprimento do SLA também apresentam menor taxa de resolução.
- **H6:** o canal de compra pela internet concentra parte relevante das reclamações do varejo.
- **H7:** empresas com grande volume não são necessariamente as de pior desempenho relativo.

> As hipóteses serão testadas com estatística descritiva, intervalos de confiança e testes de associação/comparação adequados aos tipos e à distribuição das variáveis.

## 10. Riscos e cuidados metodológicos

1. **Representatividade:** a plataforma reúne somente empresas participantes e consumidores que decidiram registrar uma reclamação; os resultados não representam automaticamente todo o mercado.
2. **Volume não é taxa:** mais reclamações podem refletir maior base de clientes. Sem denominador de clientes ou transações, não será calculada uma taxa real de reclamação por empresa.
3. **Notas ausentes:** satisfação existe apenas quando o consumidor avalia; a ausência pode não ser aleatória.
4. **Solução oficial versus confirmada:** a metodologia oficial contabiliza reclamações não avaliadas como resolvidas, exigindo apresentação separada da resolução observada.
5. **Comparabilidade:** empresas de segmentos e portes diferentes não devem ser comparadas sem filtros e amostra mínima.
6. **Privacidade:** serão usadas somente variáveis públicas e agregadas; nenhuma tentativa de reidentificação será realizada.
7. **Qualidade:** duplicados, datas inválidas, categorias inconsistentes, nulos e outliers serão auditados antes do cálculo dos KPIs.

## 11. Matriz de priorização proposta

Cada problema poderá receber um **Índice de Prioridade CX**, calculado somente após a validação das distribuições:

$$Prioridade = w_1V + w_2R + w_3S + w_4I$$

Todos os componentes serão normalizados antes da combinação. Os pesos serão transparentes e o dashboard permitirá visualizar os indicadores originais para que o índice não esconda o contexto. Inicialmente, serão testados pesos iguais; qualquer ponderação diferente precisará de justificativa de negócio.
